Visualization 1: Conference win percentage vs. tournament advancement 

This is relevant to the analysis because it reflects the relationship between conference win percentage and tournament advancement to determine whether performing better during conference season can be associated with advancing further in the NCAA tournament. It can also reveal exceptions such as teams with strong conference records that are eliminated early. These patterns can help determine whether conference performance is a useful predictor of tournament advancement. 

```r
options(repr.plot.width = 10, repr.plot.height = 10)
Womens_data <- read_csv(url)|> 
    drop_na()
Visualization_1 <- Womens_data |>
    ggplot(aes(x = tourney_finish, y = conf_wins_pct))+
    geom_point()+
    labs(tittle = "Conference Win Percentage and Tournament Advancement", 
         x = "Conference Win Percentage", 
         y = "Tournament Finish")+
  theme(text = element_text(size=16)) 
Visualization_1
```

Visualization 2: First-game location vs. Tournament advancement 

This is relevant to the analysis because it examines whether a team's first-game location (home or away) is associated with its overall tournament advancement. First game location is a unique predictor that looks beyond measuring the teams performance and seeks to determine whether external circumstances under which a team begins the tournament may influence its success. By analyzing teams playing at their home court it reveals external advantages such as crowd support and familiarity with the environment. This could potentially improve the teams advancement because of the confidence gained by the home advantage. 

```r
Visualization_2 <- Womens_data |>
    ggplot(aes(x=first_game_at_home)) +
    geom_bar(aes(fill = tourney_finish), position = "fill")+
    labs(title = "Tournament Advancement by First Game location at home court (Yes or No)", 
        x = "First Game location at home",
        y = "Proportion of Teams",
        fill = "Tournament Finish") +
theme(text = element_text(size=16)) 
Visualization_2
```
